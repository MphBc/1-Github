การจำลองข้อมูล

```python
from sklearn.linear_model import LinearRegression
```

```python
x = rng.rand(50)*10
y = 2*x + rng.randn(50)

#Linear regression model
model = LinearRegression()

x_new = x.reshape(-1,1) #จัดการข้อมูลใหม่

#Train
model.fit(x_new,y)

print(model.score(x_new,y)) #ค่า R-square
print(model.coef_) #ค่า Coefficient
print(model.intercept_) #ค่า intercept

#Test model
xfit = np.linspace(-1,11)
xfit_new = xfit.reshape(-1,1)
print(xfit_new.shape) # Check ขนาดของข้อมูล

yfit = model.predict(xfit_new)

#Analysis model
plt.scatter(x,y)
plt.plot(xfit,yfit)
plt.show()
```

# ML Predict Weather

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
```

```python
# Example for a CSV file
url = "https://raw.githubusercontent.com/kongruksiamza/MachineLearning/master/Linear%20Regression/Weather.csv"
dataset = pd.read_csv(url)
```

```python
dataset.columns
```

https://chatgpt.com/share/6948ea7a-8e88-800a-a349-c7765420a66d
ไม่ต้องแปลง 1มิติ 2มิติ เสมอไป

```python
# Train & Test set
x = dataset["MinTemp"].values.reshape(-1,1) #เปลี่ยนจาก 1มิติ เป็น 2มิติ
y = dataset["MaxTemp"].values.reshape(-1,1)

# 80% / 20%
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=0) #นำข้อมูลมาแบ่งสัดส่วนการเรียนรู้ให้ model

#Training
model = LinearRegression()
model.fit(x_train,y_train)

#Test
y_predict = model.predict(x_test) #ข้อมูล Temp ต่ำสุด (x_test) จะทำนาย Temp สูงสุดออกมาเท่าไร

plt.scatter(x_test,y_test)
plt.plot(x_test,y_predict,color="red",linewidth=2)
plt.show()
```

# เทียบข้อมูลจริง กับ ข้อมูลที่ทำนาย ได้

```python
#Compare true data and presict data
df = pd.DataFrame({'Actually':y_test.flatten(),'Predicted':y_predict.flatten()}) #แปลงข้อมูล จาก 2มิติ เป็น 1มิติ
print(df.head())
```

```python
df1 = df.head(20)
df1.plot(kind="bar",figsize=(16,10))
plt.show()
```

# การวัดผล Model

```python
from sklearn import metrics
```

```python
print("MAE = " ,metrics.mean_absolute_error(y_test,y_predict))
print("MSE = " ,metrics.mean_squared_error(y_test,y_predict))
print("RMSE = " ,np.sqrt(metrics.mean_squared_error(y_test,y_predict)))
print("Score = ",metrics.r2_score(y_test,y_predict))
```

ไม่ค่อย แม่นยำเท่าไร Score 76.7%

#BinarySGD

```python
from scipy.io import loadmat
from sklearn.datasets import fetch_openml #load data
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import SGDClassifier #สร้าง Model
from sklearn.model_selection import cross_val_score #วัดผล
```

```python
# This replaces the need for the .mat file
mnist_raw = fetch_openml('mnist_784', version=1, as_frame=False)
```

สร้าง dictionary

```python
mnist = {
    "data":mnist_raw["data"],
    "target":mnist_raw["target"].flatten().astype('int64')
}
```

```python
# Variable
x,y = mnist["data"],mnist["target"]
```

```python
# Training => 1-60,000
# Test => 60,001 - 70,000
x_train, x_test = x[:60000], x[60000:] # Train 60K ชุด
y_train, y_test = y[:60000], y[60000:]
```

```python
# Check data
print(x_train.shape)
print(x_test.shape)
print(y_train.shape)
print(y_test.shape)
```

## Binary Classification

จะดูว่าในตำแหน่งที่ 5000 เป็น 0 หรือไม่

```python
# แยก Class 0, ไม่ใช่ Class 0
# ข้อมูล 1 ค่า -> model -> Class 0 หรือไม่? True:False
predict_number = 5000
y_train_0 = (y_train == 0)
y_test_0 = (y_test == 0)

print(y_train_0)
print(y_test_0)
```

## สร้าง Model

```python
def displayImage(x):
  plt.imshow(
  x.reshape(28,28), #ขนาดรูปภาพ
  cmap=plt.cm.binary,
  interpolation="nearest")
  plt.show()

def displayPredict(clf, actually_y, x):
  x2 = x.reshape(1,-1)
  print("Actually = ", actually_y)
  print("Prediction = ", clf.predict(x2)[0]) #การทำนายผล
  displayImage(x)
```

```python
sgd_clf = SGDClassifier()
sgd_clf.fit(x_train,y_train_0)
```

```python
displayImage(x_test[predict_number])
```

```python
displayPredict(sgd_clf,y_test[predict_number],x_test[predict_number])
```

cv = 3 => ค่า k-fold คือการแบ่งข้อมูลออกเป็น 3 ครั้ง หรือการทดลอง 3 ครั้ง

```python
score =  cross_val_score(sgd_clf,x_train,y_train_0,cv=3,scoring="accuracy")
print(score)
```
